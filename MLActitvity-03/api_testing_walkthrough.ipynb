# ML Prediction API Testing Walkthrough\nThis notebook documents a practical API test workflow for the Iris FastAPI service.

## 1. Start the API\nRun in a terminal: `uvicorn app.main:app --reload`. Then visit `http://127.0.0.1:8000/docs`.

```python
import requests\nBASE_URL = 'http://127.0.0.1:8000'\nrequests.get(f'{BASE_URL}/health').json()
```

## 2. Send a prediction request

```python
payload = {\n  'sepal_length_cm': 5.1,\n  'sepal_width_cm': 3.5,\n  'petal_length_cm': 1.4,\n  'petal_width_cm': 0.2\n}\nresponse = requests.post(f'{BASE_URL}/predict', json=payload)\nresponse.status_code, response.json()
```

## 3. Test validation\nRemove a required field or send an invalid value to confirm the API responds with HTTP 422.

```python
invalid_payload = {'sepal_length_cm': 5.1, 'sepal_width_cm': 3.5, 'petal_length_cm': 1.4}\ninvalid_response = requests.post(f'{BASE_URL}/predict', json=invalid_payload)\ninvalid_response.status_code, invalid_response.json()
```
